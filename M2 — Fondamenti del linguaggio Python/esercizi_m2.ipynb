## ESERCIZIO 1 M2

```python
secondi_totali = 9296
ore = secondi_totali // 3600
minuti = (secondi_totali % 3600) // 60
secondi = secondi_totali % 60
print(f"{ore}:{minuti}:{secondi}")
ore_decimali = secondi_totali / 3600
print(f"in ore decimali: {ore_decimali:.2f}")
```

## ESERCIZIO 2 

```python
nome = "Chiara" 
voto1 = 7.5
voto2 = 8
voto3 = 6.5
assenze = 12
media = ((voto1 + voto2 + voto3) / 3)
if assenze > 50:
    giudizio = "non valutabile"
elif media >= 8:
    giudizio = "ottimo"
elif media >= 6:
    giudizio = "sufficiente"
else:
    giudizio = "insufficiente"


print(f"{nome}: media {media:.2f}, giudizio {giudizio}") 
```

## ESERCIZIO 3 

```python
valore_iniziale = int(input("inserire il valore iniziale"))
valore_finale = int(input("inserire il valore finale"))
passo = int(input("inserire il passo"))


if passo <= 0:
    print("Errore il passo deve essere positivo")
else:
    print(f"{'km':>5} {'miglia':>7} {'metri':>6}")
    print("-" * 21)

for km in range(valore_iniziale, valore_finale + 1, passo):
        miglia = km * 0.621371
        metri = km * 1000


        print(f"{km:>5} {miglia:>7.3f} {metri:>6}")
    
```

## ESERCIZIO 5

```python
frase = "Il Laboratorio 4BI dura 22 ore"

vocali_totali = 0
consonanti_totali = 0
cifre_totali = 0
altri_totali = 0


vocali = "aeiou"
consonanti = "bcdfghljhkmnpqrstvwxyz"
for carattere in frase.lower():
    if carattere.isdigit():
        cifre_totali += 1
    elif carattere in vocali:
        vocali_totali += 1
    elif carattere in consonanti:
        consonanti_totali += 1
    else:
        altri_totali += 1

print(f"vocali: {vocali_totali}, consonanti: {consonanti_totali}, cifre: {cifre_totali}, altri: {altri_totali}")

totale_caratteri = len(frase)
percentuale_vocali = 0.0

if totale_caratteri > 0:
 percentuale_vocali = (vocali_totali / totale_caratteri) * 100
print(f"percentuale di vocali: {percentuale_vocali:.1f}%")
    
```
